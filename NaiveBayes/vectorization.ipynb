```python
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('spam_dataset.csv')
```

```python
df.head()
```

```python
df['Category'] = df['Category'].replace({'ham':'Not Spam'})
```

```python
df['Category'].unique()
```

```python
x = df[['Message']]
y = df['Category']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
vectorizer = TfidfVectorizer(stop_words='english',binary=False,max_features=100,min_df=1,max_df=0.8,ngram_range=(2,2))
```

```python
vector = vectorizer.fit_transform(xtrain['Message'])
```

```python
vector.toarray()
```
