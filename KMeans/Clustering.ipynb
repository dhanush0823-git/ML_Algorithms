```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import seaborn as sns
```

```python
df = pd.read_csv('Mall_Customers.csv')
```

```python
df.head()
```

```python
df['Gender'] = df['Gender'].map({'Male':1,'Female':0})
```

```python
features = df.drop(columns=['CustomerID'])
```

```python
df.head()
```

```python
model = KMeans(n_clusters=4,random_state=42)
```

```python
model.fit(features)
```

```python
model.predict(features)
```

```python
df['result'] = model.fit_predict(features)
```

```python
# to find the best cluster values We use ELBOW-Method

scores = []
for i in range(2,11):
    model = KMeans(n_clusters=i,random_state=42)
    model.fit(features)
    #inertia_
    scores.append(model.inertia_)
print(scores)
```

```python
sns.lineplot(x = range(2,11),y = scores,markers='o')
```

## We should see where the curve starts and that is the **BEST --> n_cluster**

### **4 is the Best n_clusters**

```python
model = KMeans(n_clusters=4,random_state=42)
clusters = model.fit_predict(features)
```
